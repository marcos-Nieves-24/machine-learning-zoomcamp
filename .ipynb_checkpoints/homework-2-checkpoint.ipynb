```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
%matplotlib inline
```

```python
# importing data
data = pd.read_csv("data/car-pricing/data.csv")
data.head()
```

```python
# functions for the preprocessing
# normalizing titles
def normalizing_colname(df):
    df= df.copy()
    df.columns = df.columns.str.lower().str.replace(' ', '_')
    colname = list(df.dtypes[df.dtypes == 'str'].index)
    
    for col in colname:
        df[col] = df[col].str.lower().str.replace(' ', '_')
    return df

# function for spliting data
def split_data(df):
    df = df.copy()
    
    n = len(df)
    n_val = int(n * 0.2)
    n_test = int(n * 0.2)
    n_train = n - n_val - n_test

    idx = np.arange(n)
    np.random.seed(2)
    np.random.shuffle(idx)

    data_train = df.iloc[idx[:n_train]]
    data_val = df.iloc[idx[n_train:n_train+n_val]]
    data_test = df.iloc[idx[n_train+n_val:]]
    
    data_train = data_train.reset_index(drop=True) 
    data_val = data_val.reset_index(drop=True) 
    data_test = data_test.reset_index(drop=True) 

    return data_train, data_val, data_test
    
def y_train_log_col(df_train, df_val, df_test, colname):
    df_train = df_train.copy()
    df_val = df_val.copy()
    df_test = df_test.copy()
    
    y_train = np.log1p(df_train[colname].values)
    y_val = np.log1p(df_val[colname].values)
    y_test = np.log1p(df_test[colname].values)
        
    return (y_train, y_val, y_test)

# prepares data for linear regression model
def prepare_X(df):
    df = df.copy()
    numerical_variables = [
    'engine_hp',
    'engine_cylinders',
    'highway_mpg',
    'city_mpg',
    'popularity']
    categorical_variables = [
    'engine_fuel_type',
    'driven_wheels',
    'market_category',
    'vehicle_size',
    'vehicle_style']

    features = numerical_variables.copy()
    
    df['age'] = 2017 - df.year
    features.append('age')

    for num in [2, 3, 4]:
        df['num_doors_%s' % num] = (df.number_of_doors == num).astype('int')
        features.append('num_doors_%s' % num)
    
    categories = {}
    for c in categorical_variables:
        categories[c] = list(df[c].value_counts().head().index)
    
    for c, values in categories.items():
        for v in values:
            df['%s_%s' % (c, v)] = (df[c] == v).astype('int')
            features.append('%s_%s' % (c, v))
            
    data_num = df[features]
    data_num = data_num.fillna(0)
    X = data_num.values
    return X

# Create a vector-vector multiplication linear regression
def dot(xi, w):
    n = len(xi)
    res = 0.0
    for j in range(n):
        res = res + xi[j] * w[j]
    return res
# train a linear regression model
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    XTX.dot(XTX_inv)

    w_full = XTX_inv.dot(X.T).dot(y)
    return w_full[0], w_full[1:] 
# calculate rmse for the model
def rmse(y, y_pred):
    se = (y - y_pred) ** 2
    mse = se.mean()
    return np.sqrt(mse)
```

```python
# normalizing colnames
data_clean = normalizing_colname(data)
data_clean
```

```python
# Splitting data 
data_train, data_val, data_test = split_data(data_clean)
```

```python
# extract respons/price and normalizing 
y_train, y_val, y_test = y_train_log_col(data_train, data_val, data_test, 'msrp')
```

```python
# Deleting MSRP column 

del data_train['msrp']
del data_val['msrp']
del data_test['msrp']
```

```python
# linear regression
X_train = prepare_X(data_train)
w0, w = train_linear_regression(X_train, y_train)

X_val = prepare_X(data_val)
y_pred = w0 + X_val.dot(w)

rmse(y_val, y_pred)
```
